# Data Pre-processing (pipeline)

Import libraries/packages + split data

```python
import sys

sys.path.append("../")
from src.data_utils import get_feature_lists
from src.config import SEED, BASE_PATH
from src.preprocess import transform_export_data, SelectiveDropper
import pandas as pd
import warnings
import numpy as np

# Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import (
    OneHotEncoder,
    OrdinalEncoder,
    MinMaxScaler,
    FunctionTransformer,
)
from sklearn.pipeline import Pipeline
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer

df_import = pd.read_parquet(BASE_PATH / "data" / "raw" / "Cleaned_ORN.parquet")
X_df = df_import.drop("ORN", axis=1)
y_df = df_import["ORN"]
```

Classify features by data type

```python
##Imported func from src
feature_lists = get_feature_lists(X_df)
binary_cols = feature_lists["Binary"]
numerical_cols = feature_lists["Numerical"]
nominal_cols = feature_lists["Nominal"]
ordinal_cols = feature_lists["Ordinal"]
```

```python
################# Shared pipeline #################
nominal_pipeline = Pipeline(
    [("encoder", OneHotEncoder(handle_unknown="ignore", sparse_output=False))]
)
no_exp_cols = ["PLATETIME", "MEDUSED", "SURGUSED"]
################# Base-specific pipeline #################
base_num_pipeline = Pipeline(
    [
        (
            "imputer",
            IterativeImputer(
                estimator=None,  # default = BayesianRidge
                initial_strategy="median",
                max_iter=10,
                sample_posterior=False,  # deterministic
                random_state=SEED,
            ),
        ),
        ("scaler", MinMaxScaler()),
    ]
)
################# Nomogram-specific pipeline #################
nomo_num_pipeline = Pipeline(
    [
        (
            "imputer",
            IterativeImputer(
                estimator=None,  # default = BayesianRidge
                initial_strategy="median",
                max_iter=10,
                sample_posterior=False,  # deterministic
                random_state=SEED,
            ),
        ),
    ]
)
################# Combine all preprocessing #################
base_preprocessor = Pipeline(
    [
        (
            "column_transform",
            ColumnTransformer(
                [
                    ("num", base_num_pipeline, numerical_cols),
                    ("cat", nominal_pipeline, nominal_cols),
                    ("other", "passthrough", binary_cols + ordinal_cols),
                ]
            ),
        ),
        ("drop_redundant", SelectiveDropper(no_exp_cols)),
    ]
)
base_preprocessor.set_output(transform="pandas")
nomo_preprocessor = Pipeline(
    [
        (
            "column_transform",
            ColumnTransformer(
                [
                    ("num", nomo_num_pipeline, numerical_cols),
                    ("cat", nominal_pipeline, nominal_cols),
                    ("other", "passthrough", binary_cols + ordinal_cols),
                ]
            ),
        ),
        ("drop_redundant", SelectiveDropper(no_exp_cols)),
    ]
)
pipeline_dict = {"base": base_preprocessor, "nomo": nomo_preprocessor}
nomo_preprocessor.set_output(transform="pandas")
```

```python
data_path = BASE_PATH / "data" / "processed"
pipeline_path = BASE_PATH / "data" / "pipelines"
split_data = {}
for process_type, pipeline in pipeline_dict.items():
    print(f"{process_type}...")
    nomo = True if process_type == "nomo" else False
    split_data[process_type] = transform_export_data(
        X=X_df,
        y=y_df,
        nomo=nomo,
        preprocessor=pipeline,
        data_path=data_path,
        pipeline_path=pipeline_path,
    )
```
